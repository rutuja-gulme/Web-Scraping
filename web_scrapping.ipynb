# Web Scraping with Python

This project demonstrates how to extract structured information from a webpage using Python.

### Tools Used

* Python
* Requests
* BeautifulSoup
* LXML
* Pandas

### Objective

To fetch webpage content, parse HTML elements, and extract useful information into a structured format.


### Scraping Considerations

* Check the website's Terms and Conditions before scraping.
* Avoid sending too many requests in a short period.
* Monitor scraping activity and handle requests responsibly.
* Clean and preprocess the extracted data before analysis.
* Configure the scraper according to the website structure.


```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

## Target Website

The project extracts archive information from Student News Daily.

**Source:** Student News Daily — Daily News Article Archive


```python
url = 'https://www.studentnewsdaily.com/archive/daily-news-article/'
```

## Fetch and Parse Webpage

The webpage is accessed using the Requests library, and BeautifulSoup is used to parse the HTML content.


```python
response = requests.get(url, timeout=10)
response.raise_for_status()

# Parse the HTML content
soup = BeautifulSoup(response.content, "lxml")
```

```python
content = soup.find("div", class_="col-md-10 col-lg-10 col-xl-10")

print(content is not None)
```

## Data Extraction

The scraper extracts article information from the webpage and stores the results in a structured Pandas DataFrame.

```python
articles = content.find_all("p")

for article in articles[:10]:
    print(article.get_text(" ", strip=True))
```

```python
data = []

for article in articles:
    text = article.get_text(" ", strip=True)
    data.append(text)

df = pd.DataFrame(data, columns=["article"])

df.head(10)
```

```python
df[["date", "article_title"]] = df["article"].str.split(" - ", n=1, expand=True)

df = df[["date", "article_title"]]

df.head(10)
```

```python
df["date"] = pd.to_datetime(df["date"])

df.info()
```

```python
df.isnull().sum()
```

```python
print(f"Total articles scraped: {len(df)}")
```

## Exported Dataset

The cleaned and structured data is exported as a CSV file for further analysis or reuse.

```python
df.to_csv("scraped_articles.csv", index=False)

print("CSV file saved successfully!")
```
